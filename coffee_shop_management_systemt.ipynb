```python
products = {
    "Coffee": {
        "price": 120,
        "stock": 50
    },

    "Tea": {
        "price": 30,
        "stock": 100
    },

    "Burger": {
        "price": 180,
        "stock": 30
    },

    "Pizza": {
        "price": 250,
        "stock": 20
    },

    "Sandwich": {
        "price": 150,
        "stock": 25
    }
}

print("Products loaded successfully!")
```

```python
def view_menu():

    print("\n========================================")
    print("              COFFEE MENU")
    print("========================================")

    for name, details in products.items():

        print(
            f"{name:<10} "
            f"Price: ₹{details['price']:<6} "
            f"Stock: {details['stock']}"
        )


view_menu()
```

```python
def add_product():

    print("\n========================================")
    print("             ADD PRODUCT")
    print("========================================")

    name = input("Enter product name: ").strip()

    if name in products:
        print("Product already exists!")
        return

    try:
        price = float(input("Enter product price: "))
        stock = int(input("Enter product stock: "))

        if price <= 0:
            print("Price must be greater than 0!")
            return

        if stock < 0:
            print("Stock cannot be negative!")
            return

        products[name] = {
            "price": price,
            "stock": stock
        }

        print(f"{name} added successfully!")

    except ValueError:
        print("Please enter valid numbers!")


add_product()
```

```python
def search_product():

    print("\n========================================")
    print("            SEARCH PRODUCT")
    print("========================================")

    name = input("Enter product name to search: ").strip()

    if name in products:

        details = products[name]

        print("\nProduct Found! ✅")
        print("------------------------")
        print("Product :", name)
        print("Price   : ₹", details["price"])
        print("Stock   :", details["stock"])

    else:
        print("\nProduct not found! ❌")


search_product()
```

```python
def update_product():

    print("\n========================================")
    print("           UPDATE PRODUCT")
    print("========================================")

    name = input("Enter product name to update: ").strip()

    if name not in products:
        print("Product not found! ❌")
        return

    print("\nCurrent Details:")
    print("Product :", name)
    print("Price   :", products[name]["price"])
    print("Stock   :", products[name]["stock"])

    try:
        new_price = float(input("\nEnter new price: "))
        new_stock = int(input("Enter new stock: "))

        if new_price <= 0:
            print("Price must be greater than 0!")
            return

        if new_stock < 0:
            print("Stock cannot be negative!")
            return

        products[name]["price"] = new_price
        products[name]["stock"] = new_stock

        print("\nProduct updated successfully! ✅")

    except ValueError:
        print("Please enter valid numbers! ❌")


update_product()
```

```python
def delete_product():

    print("\n========================================")
    print("           DELETE PRODUCT")
    print("========================================")

    name = input("Enter product name to delete: ").strip()

    if name not in products:
        print("Product not found! ❌")
        return

    print("\nProduct Details:")
    print("Product :", name)
    print("Price   :", products[name]["price"])
    print("Stock   :", products[name]["stock"])

    confirm = input("\nAre you sure you want to delete? (yes/no): ").lower()

    if confirm == "yes":

        del products[name]

        print("\nProduct deleted successfully! ✅")

    else:
        print("\nDelete operation cancelled.")


delete_product()
```

```python
def place_order():

    print("\n========================================")
    print("              PLACE ORDER")
    print("========================================")

    order = []
    subtotal = 0

    while True:

        view_menu()

        name = input("\nEnter product name (or 'done' to finish): ").strip()

        if name.lower() == "done":
            break

        if name not in products:
            print("Product not found! ❌")
            continue

        try:
            quantity = int(input("Enter quantity: "))

            if quantity <= 0:
                print("Quantity must be greater than 0!")
                continue

            if quantity > products[name]["stock"]:
                print("Insufficient stock! ❌")
                print("Available stock:", products[name]["stock"])
                continue

            price = products[name]["price"]
            total = price * quantity

            order.append({
                "name": name,
                "price": price,
                "quantity": quantity,
                "total": total
            })

            subtotal += total

            products[name]["stock"] -= quantity

            print(f"{quantity} {name} added to cart! ✅")

        except ValueError:
            print("Please enter a valid quantity!")

    if not order:
        print("\nNo items ordered.")
        return

    print("\n========================================")
    print("              ORDER SUMMARY")
    print("========================================")

    for item in order:
        print(
            f"{item['name']} | "
            f"{item['quantity']} x ₹{item['price']} "
            f"= ₹{item['total']}"
        )

    print("----------------------------------------")
    print(f"Subtotal : ₹{subtotal}")

    gst = subtotal * 0.05
    grand_total = subtotal + gst

    print(f"GST 5%   : ₹{gst:.2f}")
    print("----------------------------------------")
    print(f"TOTAL    : ₹{grand_total:.2f}")

    print("\nOrder placed successfully! ✅")

    return order, grand_total


```

```python
from datetime import datetime

bill_number = 1001


def generate_bill(order, grand_total):

    global bill_number

    print("\n")
    print("=" * 50)
    print("             ☕ COFFEE SHOP")
    print("              CUSTOMER BILL")
    print("=" * 50)

    customer_name = input("Enter customer name: ").strip()

    current_date = datetime.now().strftime("%d-%m-%Y")
    current_time = datetime.now().strftime("%I:%M %p")

    print(f"\nBill No.      : {bill_number}")
    print(f"Customer Name : {customer_name}")
    print(f"Date          : {current_date}")
    print(f"Time          : {current_time}")

    print("-" * 50)
    print(f"{'Product':<15}{'Qty':<8}{'Price':<12}{'Total':<10}")
    print("-" * 50)

    subtotal = 0

    for item in order:

        print(
            f"{item['name']:<15}"
            f"{item['quantity']:<8}"
            f"₹{item['price']:<11.2f}"
            f"₹{item['total']:<10.2f}"
        )

        subtotal += item["total"]

    gst = subtotal * 0.05

    print("-" * 50)
    print(f"{'Subtotal':<35} ₹{subtotal:.2f}")
    print(f"{'GST (5%)':<35} ₹{gst:.2f}")
    print("-" * 50)
    print(f"{'Grand Total':<35} ₹{grand_total:.2f}")
    print("=" * 50)
    print("        Thank You! Visit Again ☕")
    print("=" * 50)

    bill_number += 1
```

```python
result = place_order()

if result:
    order, grand_total = result
    generate_bill(order, grand_total)
```

```python
sales = []

print("Sales system ready! ✅")
```

```python
def save_sale(order, grand_total):

    sale = {
        "order": order,
        "total": grand_total
    }

    sales.append(sale)

    print("Sale saved successfully! ✅")
```

```python
def sales_report():

    print("\n========================================")
    print("             SALES REPORT")
    print("========================================")

    if not sales:
        print("No sales available.")
        return

    total_sales = 0

    print("Total Orders :", len(sales))

    for i, sale in enumerate(sales, start=1):

        print(
            f"Order {i:<5} : ₹{sale['total']:.2f}"
        )

        total_sales += sale["total"]

    print("----------------------------------------")
    print(f"Total Sales   : ₹{total_sales:.2f}")
    print("========================================")
```

```python
result = place_order()

if result:
    order, grand_total = result

    save_sale(order, grand_total)

    generate_bill(order, grand_total)
```

```python
sales_report()
```

```python
def save_products():

    with open("products.txt", "w") as file:

        for name, details in products.items():

            file.write(
                f"{name},{details['price']},{details['stock']}\n"
            )

    print("Products saved successfully! ✅")


save_products()
```

```python
def load_products():

    try:

        with open("products.txt", "r") as file:

            for line in file:

                name, price, stock = line.strip().split(",")

                products[name] = {
                    "price": float(price),
                    "stock": int(stock)
                }

        print("Products loaded successfully! ✅")

    except FileNotFoundError:

        print("Products file not found! ❌")
```

```python
load_products()
```

```python
def save_sales():

    with open("sales.txt", "w") as file:

        for sale in sales:

            file.write(f"{sale['total']}\n")

    print("Sales saved successfully! ✅")
```

```python
save_sales()
```

```python
def load_sales():

    try:

        with open("sales.txt", "r") as file:

            for line in file:

                total = float(line.strip())

                sales.append({
                    "total": total
                })

        print("Sales loaded successfully! ✅")

    except FileNotFoundError:

        print("Sales file not found! ❌")
```

```python
def main_menu():

    admin_logged_in = False

    while True:

        print("\n")
        print("=" * 50)
        print("       ☕ COFFEE SHOP MANAGEMENT SYSTEM")
        print("=" * 50)

        print("1. Admin Login")
        print("2. View Menu")
        print("3. Add Product")
        print("4. Search Product")
        print("5. Update Product")
        print("6. Delete Product")
        print("7. Place Order")
        print("8. View Sales")
        print("9. Save Products")
        print("10. Exit")

        choice = input("\nEnter your choice: ").strip()

        # Admin Login
        if choice == "1":

            admin_logged_in = admin_login()

        # View Menu
        elif choice == "2":

            view_menu()

        # Add Product
        elif choice == "3":

            if admin_logged_in:
                add_product()
            else:
                print("\nPlease login as Admin first! 🔐")

        # Search Product
        elif choice == "4":

            search_product()

        # Update Product
        elif choice == "5":

            if admin_logged_in:
                update_product()
            else:
                print("\nPlease login as Admin first! 🔐")

        # Delete Product
        elif choice == "6":

            if admin_logged_in:
                delete_product()
            else:
                print("\nPlease login as Admin first! 🔐")

        # Place Order
        elif choice == "7":

            result = place_order()

            if result:

                order, grand_total = result

                save_sale(order, grand_total)

                generate_bill(order, grand_total)

        # Sales Report
        elif choice == "8":

            sales_report()

        # Save Products
        elif choice == "9":

            save_products()

        # Exit
        elif choice == "10":

            save_products()
            save_sales()

            print("\nThank you for using Coffee Shop Management System! ☕")
            print("Goodbye! 👋")

            break

        else:

            print("\nInvalid choice! Please enter a number from 1 to 10. ❌")
main_menu()
```
